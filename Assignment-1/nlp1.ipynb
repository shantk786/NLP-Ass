```python
!pip install nltk
import nltk

nltk.download('punkt')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('omw-1.4')
text = "NLTK is awesome! I love NLP :) #AI @OpenAI Let's tokenize this sentence."
print(text)
```

```python
from nltk.tokenize import WhitespaceTokenizer

wt = WhitespaceTokenizer()
whitespace_tokens = wt.tokenize(text)

print("Whitespace Tokenization:")
print(whitespace_tokens)
```

```python
from nltk.tokenize import wordpunct_tokenize

punct_tokens = wordpunct_tokenize(text)

print("\nPunctuation-based Tokenization:")
print(punct_tokens)
```
