```python
import spacy
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score

nlp = spacy.load("en_core_web_sm")
print("spaCy model loaded successfully")
```

```python
news_dataset = [
    {
        "text": """With the presentation of the Union Budget 2026, several items are expected to get cheaper while some others will become costlier for Indian consumers. The finance minister spoke about reduced customs duty on items such as cancer drugs, food for rare diseases, leather items, textile garments, lithium-ion cells, solar glass, critical minerals, aircraft manufacturing components, and overseas education. Items expected to get costlier include alcohol, cigarettes, iron ore, coal, and nuclear project components.""",
        
        # Ground-truth (manually annotated)
        "entities": [
            ("Union Budget 2026", "EVENT"),
            ("Indian", "NORP"),
            ("cancer drugs", "PRODUCT"),
            ("lithium-ion cells", "PRODUCT"),
            ("solar glass", "PRODUCT"),
            ("alcohol", "PRODUCT"),
            ("cigarettes", "PRODUCT"),
            ("iron ore", "PRODUCT"),
            ("coal", "PRODUCT")
        ]
    }
]
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

for data in news_dataset:
    doc = nlp(data["text"])
    print("Predicted Entities:")
    for ent in doc.ents:
        print(ent.text, "->", ent.label_)
```

```python
y_true = []
y_pred = []

for data in news_dataset:
    text = data["text"]
    true_entities = data["entities"]
    
    doc = nlp(text)
    predicted_entities = [(ent.text, ent.label_) for ent in doc.ents]
    
    for true_ent in true_entities:
        y_true.append(true_ent[1])
        if true_ent in predicted_entities:
            y_pred.append(true_ent[1])
        else:
            y_pred.append("O")   # O = entity missed
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_true_enc = le.fit_transform(y_true)
y_pred_enc = le.transform(y_pred)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_true_enc, y_pred_enc)
precision = precision_score(y_true_enc, y_pred_enc, average="macro", zero_division=0)
recall = recall_score(y_true_enc, y_pred_enc, average="macro", zero_division=0)
f1 = f1_score(y_true_enc, y_pred_enc, average="macro", zero_division=0)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-score :", f1)
```
