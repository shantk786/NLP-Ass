```python
!pip install nltk
import nltk
from nltk.corpus import wordnet as wn

nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
text_words = ["good", "bad", "car", "dog"]
```

```python
def get_synonyms(word):
    synonyms = set()
    for syn in wn.synsets(word):
        for lemma in syn.lemmas():
            synonyms.add(lemma.name())
    return synonyms

print("Synonyms of 'good':")
print(get_synonyms("good"))
```

```python
{'good', 'beneficial', 'right', 'upright', 'honorable', ...}
```

```python
def get_antonyms(word):
    antonyms = set()
    for syn in wn.synsets(word):
        for lemma in syn.lemmas():
            if lemma.antonyms():
                antonyms.add(lemma.antonyms()[0].name())
    return antonyms

print("Antonyms of 'good':")
print(get_antonyms("good"))
```

```python
def get_hypernyms(word):
    hypernyms = set()
    for syn in wn.synsets(word):
        for hyper in syn.hypernyms():
            hypernyms.add(hyper.name())
    return hypernyms

print("Hypernyms of 'dog':")
print(get_hypernyms("dog"))
```

```python
for word in text_words:
    print(f"\nWord: {word}")
    print("Synonyms :", list(get_synonyms(word))[:5])
    print("Antonyms :", list(get_antonyms(word)))
    print("Hypernyms:", list(get_hypernyms(word)))
```

```python
sentence = "The dog is a good animal"

words = sentence.lower().split()

for word in words:
    print(f"\nWord: {word}")
    print("Synonyms:", list(get_synonyms(word))[:3])
```
